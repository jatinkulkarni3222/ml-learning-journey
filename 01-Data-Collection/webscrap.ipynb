```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'} 
webpage=requests.get('https://www.ambitionbox.com/list-of-companies?campaign=homepage_companies_widget&page=1',headers=headers).text
```

```python

soup=BeautifulSoup(webpage,'lxml')
```

```python
for i in soup.find_all('h2',class_='companyCardWrapper__companyName'):
    print(i.text.strip())
    

len(soup.find_all('h2'))

```

```python
len(soup.find_all('div',class_='rating_text'))
```

```python

company=soup.find_all('div',class_='companyCardWrapper__primaryInformation')
len(company)
```

```python
len(soup.find_all('span',class_='companyCardWrapper__companyRatingCount'))
```

```python
len(soup.find_all('span',class_="companyCardWrapper__ratingValues"))
```

```python
name=[]
rating=[]
reviews=[]
for i in company:
    name.append(i.find('h2',class_='companyCardWrapper__companyName').text.strip())
    rating.append(i.find('div',class_='rating_text').text.strip())
    reviews.append(i.find('span',class_="companyCardWrapper__ratingValues"))



df=pd.DataFrame({'name':name,'ratings':rating,'review':reviews})
df
```
