```python
import pandas as pd
import requests
```

```python
response=requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=0562b53fe3009a4b7393e31197342d67&language=en-US&page=1')
```

```python
df=pd.DataFrame()
df
```

```python
for i in range(1, 7):
    response = requests.get(
        'https://api.themoviedb.org/3/movie/top_rated?api_key=0562b53fe3009a4b7393e31197342d67&language=en-US&page={}'.format(i)
    )

    df_temp = pd.DataFrame(response.json()['results'])[
        ['id', 'title', 'overview', 'release_date', 'popularity', 'vote_average', 'vote_count']
    ]

    df = pd.concat([df,df_temp], ignore_index=True)
    
```
