```python
import numpy as np
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
X,y=make_regression(n_samples=4,n_targets=1,n_features=1,n_informative=1,noise=80,random_state=13)
```

```python
plt.scatter(X,y)
plt.show()
```

```python
lr=LinearRegression()
lr.fit(X,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red')
```

#**now we will put m=78.35 and initially b=100 then calc the nearest line using gradient descent**

```python
y_pred= ((78.35*X)+100).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='green',label='b')
plt.legend()
```

```python
m=78.35
b=100
loss_slope=-2*np.sum(y-(m*X.ravel())-b)
loss_slope
```

```python
n=0.1
step_size=(n*loss_slope)
step_size
```

```python
b= b -step_size
b
```

```python
y_pred1= ((78.35*X)+b).reshape(4)
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='green',label='b={}'.format(100))
plt.plot(X,y_pred1,color='blue',label='b={}'.format(b))
plt.legend()
```

doing using iteration loop

```python
ephocs=100
m=78.35
b=-100
n=0.01

for i in range(ephocs):
  loss_slope=-2*np.sum(y-(m*X.ravel())-b)
  step_size=n*(loss_slope)
  b=b-step_size
  y_pred=m*X+b
  plt.plot(X,y_pred)

plt.scatter(X,y)
```
