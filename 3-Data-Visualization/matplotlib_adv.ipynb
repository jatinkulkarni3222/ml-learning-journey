```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.style.use('dark_background')
```

```python
subjects = ['Maths', 'English', 'Science', 'Social Studies', 'Computer']
marks = [89, 90, 45, 78, 99]
```

```python
colorss=['orange','red','green','yellow','blue']
plt.bar(subjects,marks,color=colorss,width=0.5,edgecolor='white',linewidth=2,linestyle='-.')
plt.show()
```

```python
subjects = ['Maths', 'English', 'Science', 'Social Studies', 'Computer']
marks1 = [89, 90, 45, 78, 99]
marks2 = [78, 56, 34, 90, 12]
```

```python
plt.figure(figsize=(8,8))
plt.bar(subjects,marks1)
plt.bar(subjects,marks2)
plt.xlabel('subject')
plt.ylabel('marks')
plt.show()
```

```python
subjects_len=np.arange(len(subjects))
width=0.4
```

```python
plt.figure(figsize = (8,8))

plt.bar(subjects_len, marks1, width = width,color=colorss)
plt.bar(subjects_len + width, marks2, width = width,color=colorss, alpha = 0.5)

plt.xlabel('Subjects')
plt.ylabel('Marks')

plt.show()
```
