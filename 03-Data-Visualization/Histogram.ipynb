```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.style.use('dark_background')
```

```python
marks_50_students=np.random.randint(0,100,(50))
marks_50_students
```

```python
plt.hist(marks_50_students)
plt.show()
```

```python
bin=np.arange(0,100,5)
plt.figure(figsize=(6,6))
plt.hist(marks_50_students,bins=bin,color='orange',rwidth=.6,histtype='step')
plt.xticks(np.arange(0,100,5))
plt.show()
```

```python
marks_50_students1 = np.random.randint(0, 100, (50))
marks_50_students2 = np.random.randint(0, 100, (50))
```

```python
bins=np.arange(0,100,5)
plt.hist([marks_50_students1,marks_50_students2],bins=bin,color=['orange','red'])
plt.xticks(np.arange(0,100,5))
plt.xlabel('marks')
plt.ylabel('frequency')
plt.title('marks distribution')
plt.show()
```
